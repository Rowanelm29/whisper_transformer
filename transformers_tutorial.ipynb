```python
import torch
from transformers import pipeline
```

```python
device = "cuda:0" if torch.cuda.is_available() else "cpu"
```

```python
pipe = pipeline(
  "automatic-speech-recognition",
  model="openai/whisper-tiny",
  chunk_length_s=30,
  device=device,
)
```

```python
# Transcribe an audio file (replace with your file path)
result = pipe("/audio_files/422-122949-0001.flac")

print(result["text"])
```
